## 1 lygis — Kas tu esi?

Apačioje yra keturi **kintamieji**. Kintamasis — tai vardas, kurį duodame
reikšmei, kad vėliau galėtume ja naudotis.

Dabar ten įrašyta mokytojo kortelė. Pakeiskite ją savo duomenimis ir
paleiskite langelį.

**`svajoniu_projektas`** — ką sukurtumėte, jei jau mokėtumėte viską, ko tam
reikia? Žaidimą, programėlę, robotą, svetainę? Nesvarbu, ar šiandien žinote,
kaip tai padaryti — čia rašome svajonę, ne planą.

> 💡 Tekstas rašomas tarp kabučių: `"Nuotykių programėlė"`.
> Skaičiui kabučių nereikia: `16`.
>
> 💡 Kintamųjų varduose lietuviškų raidžių nėra: `svajoniu_projektas`,
> ne `svajonių_projektas`.

````python
from kortele import rodyk_kortele, patikrink

vardas             = "Deimantas"
amzius             = 28
svajoniu_projektas = "Nuotykių programėlė"
faktas             = "Turiu dvi kates"

rodyk_kortele(vardas, amzius, svajoniu_projektas, faktas)
patikrink(1)
````

## 2 lygis — Ką mėgsti?

Vienas pomėgis kortelėje atrodytų liūdnai, todėl surašysime kelis iš karto.
Tam reikia **sąrašo** (angl. *list*) — reikšmių rinkinio laužtiniuose
skliaustuose, atskirtų kableliais.

Surašykite 2–4 savo pomėgius ir paleiskite langelį.

> 💡 Nepamirškite kablelio tarp pomėgių.

````python
from kortele import rodyk_kortele, patikrink

pomegiai = ["laipiojimas", "programavimas", "choras"]

rodyk_kortele(vardas, amzius, svajoniu_projektas, faktas, pomegiai=pomegiai)
patikrink(2)
````

## 3 lygis — Slaptas programuotojo vardas

Dabar reikšmės neįrašysime, o **apskaičiuosime** iš to, ką jau turime.

- `vardas[:3]` — pirmos trys vardo raidės
- `.upper()` — paverčia raides DIDŽIOSIOMIS
- `f"..."` — leidžia į tekstą įterpti reikšmes per `{ }`

Paleiskite ir pažiūrėkite, koks slapyvardis gavosi.

````python
from kortele import rodyk_kortele, patikrink

slapyvardis = f"{vardas[:3].upper()}-{amzius * 2}"

rodyk_kortele(vardas, amzius, svajoniu_projektas, faktas, pomegiai=pomegiai, slapyvardis=slapyvardis)
patikrink(3)
````
